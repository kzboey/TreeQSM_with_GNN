```python
import numpy as np
import matplotlib.pyplot as plt

def forward_project_2d_object(object_2d, num_angles):
    """
    Perform forward projection for a 2D object.

    Parameters:
    - object_2d: 2D numpy array representing the attenuation coefficients of the object.
    - num_angles: Number of projection angles.

    Returns:
    - sinogram: 2D numpy array representing the simulated sinogram.
    """
    num_detectors = object_2d.shape[1]
    sinogram = np.zeros((num_angles, num_detectors))

    # Define projection angles
    angles = np.linspace(0, 180, num_angles, endpoint=False)

    for i, angle in enumerate(angles):
        # Rotate the object by the current angle
        rotated_object = np.rot90(object_2d, k=int(angle / 180 * object_2d.shape[1]))

        # Perform line integral along each row to simulate X-ray projection
        projection = np.sum(rotated_object, axis=0)

        # Store the projection in the sinogram
        sinogram[i, :] = projection

    return sinogram

# Create a simple 2D object (e.g., a rectangle)
object_size = 64
object_2d = np.zeros((object_size, object_size))
object_2d[object_size // 4:3 * object_size // 4, object_size // 4:3 * object_size // 4] = 1.0

# Perform forward projection
num_angles = 180
sinogram = forward_project_2d_object(object_2d, num_angles)

# Display the original object and sinogram
plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.imshow(object_2d, cmap='gray')
plt.title('Original Object')

plt.subplot(122)
plt.imshow(sinogram, cmap='gray', aspect='auto', extent=[0, object_size, 180, 0])
plt.title('Sinogram')
plt.xlabel('Detector Position')
plt.ylabel('Projection Angle (degrees)')

plt.tight_layout()
plt.show()
```

```python
from skimage.data import shepp_logan_phantom

# Create a Shepp-Logan phantom
phantom_size = 256
shepp_logan = shepp_logan_phantom()

# Perform forward projection
num_angles = 180
sinogram_shepp_logan = forward_project_2d_object(shepp_logan, num_angles)

# Display the Shepp-Logan phantom and sinogram
plt.figure(figsize=(12, 5))
plt.subplot(121)
plt.imshow(shepp_logan, cmap='gray')
plt.title('Shepp-Logan Phantom')

plt.subplot(122)
plt.imshow(sinogram_shepp_logan, cmap='gray', aspect='auto', extent=[0, phantom_size, 180, 0])
plt.title('Sinogram')
plt.xlabel('Detector Position')
plt.ylabel('Projection Angle (degrees)')

plt.tight_layout()
plt.show()
```
