```python
class MyObject:
    def __init__(self, id, name):
        self.id = id
        self.name = name

# Create a list of objects
my_list = [MyObject(1, "Object A"), MyObject(2, "Object B"), MyObject(3, "Object C")]

# Unique parameter to search for
target_id = 4

# Use list comprehension to find the object with the matching id
result = next((obj for obj in my_list if obj.id == target_id), None)

# Check if an object with the specified id was found
if result is not None:
    print(f"Object with id {target_id} found: {result.name}")
else:
    print(f"No object with id {target_id} found.")
```
